```python
import numpy as np
import string

charcters = string.ascii_uppercase + string.digits
# print(charcters)
# Acc_no = np.random.randint(1000000,999999999)
Acc_No = ''.join(np.random.choice(list(charcters),size =15))
# print(Acc_No)
Initial_balance = 0


def Create_Account (Name,Acc_No,Age,Address,Pin):    
    print(f"Name : {Name}")
    print(f"Age : {Age}")
    print(f"Address :{Address}")
    print(f"Pin :{Pin}")
    print(f"Account no : {Acc_No}")

def Check_Balance (Balance):
    print(f"Your Balance is {Balance}")
def Deposit_Money(Amount):
    print(f"You Deposited {Amount}: ")
def Credit_Money(Amount):
    print(f"You have credited {Amount} ")

Name = str(input("Enter Your Name "))
Age = int(input("Enter your Age"))
Address = str(input("Enter your Address"))
Pin = 989945
if(Age>=18):
        Create_Account(Name,Acc_No,Age,Address,Pin)
else:
        print(f"Sorry ! You are under age You can create account after {18-Age} Years")
        # raise SystemExit
        exit()
        
# Create_Account()
print("-" * 30)

# ---- PIN check (3 attempts) ----
attempts = 3
while attempts > 0:
    entered_pin = int(input("Enter your PIN to continue: "))
    if entered_pin == Pin:
        print("PIN verified!")
        break
    else:
        attempts -= 1
        print(f"Wrong PIN. Attempts left: {attempts}")
else:
    print("Too many wrong attempts. Account locked.")
    raise SystemExit
# ---- end PIN check ----

print("Click 1 to Check balance ")
print("Click 2 to Deposit Money")
print("Click 3 to Withdraw Money ")
print("Click 4 to Exit the loop ")
print("-" * 30)


# i = 0 
while True:
        Choice = int(input("Please Enter Your Choice :"))
    
        if(Choice==1):
            Check_Balance(Initial_balance)
        elif(Choice==2):
            Amount = int(input("how much you want to Deposit "))
            Initial_balance+=Amount
            Deposit_Money(Amount)
            print(f"Your final balance is {Initial_balance}")
        elif(Choice==3):
            Amount = int(input("how much you want to Credit "))
            if(Amount>Initial_balance):
                print("Low balance")
            else:
                Credit_Money(Amount)
                Initial_balance-=Amount
                print(f"your new alance is {Initial_balance}")
        elif(Choice==4):
            print("Thank you for choosing our service")
            break
        else:
            print("invalid choice")
    
```
